```python
import plotly.express as px
```

```python
df_nivel_usd = df_tratado.groupby('nivel')['usd'].mean().reset_index()
ordem_senioridade = ['Junior', 'Pleno', 'Senior', 'Executivo']
df_nivel_usd['nivel'] = pd.Categorical(df_nivel_usd['nivel'], categories=ordem_senioridade, ordered=True)
df_nivel_usd = df_nivel_usd.sort_values('nivel')

fig = px.bar(df_nivel_usd,
             x='nivel',
             y='usd',
             title='Salário médio por nível de senioridade',
             labels={'nivel': 'Nível de senioridade', 'usd': 'Salário médio anual em USD'},
             color='nivel'  # Color by experience level
            )

fig.show()
```

```python
remoto_contagem = df_tratado['remoto'].value_counts().reset_index()
remoto_contagem.columns = ['remoto', 'quantidade']

fig = px.pie(remoto_contagem,
             names='remoto',
             values='quantidade',
             title='Distribuição de trabalho remoto',
             )

fig.show()
```

```python
fig = px.pie(remoto_contagem,
             names='remoto',
             values='quantidade',
             title='Distribuição de trabalho remoto',
             hole=0.5
             )

fig.update_traces(textinfo='percent+label')

fig.show()
```

```python
df_data_scientist = df_tratado[df_tratado['cargo'] == 'Data Scientist']

df_pais_usd = df_data_scientist.groupby('localizacao')['usd'].mean().reset_index()
df_pais_usd = df_pais_usd.sort_values('usd', ascending=False)

fig = px.bar(df_pais_usd,
             x='localizacao',
             y='usd',
             title='Salário médio de Cientista de Dados por País',
             labels={'localizacao': 'País', 'usd': 'Salário médio anual em USD'},
             color='localizacao'
            )

fig.show()
```

```python
import pycountry

#Função para converter ISO-2 ara ISO-3
def iso2_to_iso3(code):
  try:
    return pycountry.countries.get(alpha_2=code).alpha_3
  except:
    return None

#Criar nova coluna com código ISO-3
df_tratado['residencia_iso3'] = df_tratado['residencia'].apply(iso2_to_iso3)

#Calcular média salarial por país (ISO-3)
df_ds = df_tratado[df_tratado['cargo'] == 'Data Scientist']
media_ds_pais = df_ds.groupby('residencia_iso3')['usd'].mean().reset_index()

#Gerar o mapa
fig = px.choropleth(media_ds_pais,
                    locations='residencia_iso3',
                    color='usd',
                    color_continuous_scale='rdylgn',
                    title='Média salarial de Cientista de Dados por País',
                    labels={'usd': 'Salário médio anual em USD', 'residencia_iso3': 'País'}
                   )

fig.show()
```
